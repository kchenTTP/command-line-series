### **Terminal vs GUI**

In the past, computers only had text-based interfaces. Today, most use a graphical user interface (GUI). To run text commands, you need a program that gives you a place to type them. That program is the **terminal**, short for **terminal emulator** (also called the **command line**). It imitates those old text devices and gives you a window where you can type commands to the computer.

### **Terminal, Shell, Prompt**

A terminal setup has 3 main parts.

- **Terminal**: the window you type into. It shows text and sends your keystrokes to the shell.
- **Shell**: the program that reads and runs your commands. Common ones today are `bash`, `zsh`, and `fish`.
- **Prompt**: the short text that shows the shell is ready. It often shows your user name, your machine, and your current folder.

> 💡 You can customize the prompt so it might look different on different systems


````
# Run this in the terminal to see which shell you're using
%%bash
echo $SHELL
````

### **Where do commands live?**

Commands like `echo` aren't magic. Each one is a program stored somewhere on the computer. The shell has a list of folders it checks when you type a name. Use `which` to see where a command was found:

## **Command Anatomy**

Before we learn more commands, let's learn how to read any command. Most look like this:

```sh
  ls       -l       -h       folder/
   ↑        ↑        ↑         ↑
command   option   option   argument
```

- **Command**: what to do
- **Options** (or **flags**): change how the command behaves. Often starts with `-`
- **Arguments**: what the command should act on

> 📒 **Note**: Spaces separate the parts, so **spaces matter**. `ls -l` is not the same as `ls -l`.


`ls` is a command that lists the files and folders in your current directory. Let's try it out with different options. First, no options:


````
%%bash
ls
````

Add `-l` for a long listing to see permissions, size, date:


````
%%bash
ls -l
````

Add `-h` to show sizes in human-readable form (KB, MB). Short options can be combined: `-l -h` is the same as `-lh`.


````
%%bash
ls -lh
````

Now add an **argument** to look inside a specific folder:


````
%%bash
ls -lh terminal_library
````

> 📒 Two more things you'll see:
> - **Long options**: start with 2 dashes and use words: `--help`, `--version`. Many short options have a long twin.
> - Some options take a value of their own, like `head -n 3 file.txt`.

### **Paths**

A **path** is the address of a file or folder. There are 2 kinds:

- **Absolute path**: the full route from the root. It starts with `/` and works from anywhere.

```sh
/content/terminal_library/stacks/fiction/dune.txt
```

- **Relative path**: the route from the folder you are in now, not from the root.

```sh
# if you're in /content/terminal_library
stacks/fiction/dune.txt
```

<br>

#### Special shortcuts:

| Symbol | Meaning | Example |
| :-: | :-- | :-- |
| `.` | the current directory | `ls ./file.txt` |
| `..` | the parent directory (one level up) | `cd ..` |
| `/` | the root of the whole file system | `cd /` |
| `~` | your home directory | `cd ~` |
| `-` | the previous directory you were in | `cd -` |

Walking up and down the tree using these special shortcuts:


````
%%bash
cd /content/terminal_library/stacks/fiction
pwd
cd ..
pwd
cd ../../terminal_library/returns_cart
pwd
````

````
%%bash
cd /content/terminal_library/stacks/fiction
pwd
cd
pwd
cd -
````

> 📒 **Note**: Typing `cd` with nothing after it takes you home (same as `cd ~`)
